```python
from ColabTurtle.Turtle import *
initializeTurtle()

color('orange')
width(3)

# Opakuj 4x pro čtverec
for _ in range(4):
    forward(150)
    left(90)

# ## 4. Vibecoding: Pokročilé příklady
# Vibecoding je o stylu. Používáme černé pozadí, zajímavá barevná schémata a matematickou symetrii. Zkuste se inspirovat matematickými  vzory.

# ### Vzor A: The Digital Pulse (Digitální puls)
# Tento příklad využívá trik se změnou tloušťky čáry (`width`) během kreslení. Vypadá to, jako by obrazec dýchal nebo pulzoval.

# **Vysvětlení kódu:**
# *   Používáme úhel `59` stupňů (ne 60). Díky tomu se trojúhelníky nikdy přesně nepotkají a vznikne chaos.
# *   `width(i % 10 + 1)` neustále mění tloušťku čáry od 1 do 10.

from ColabTurtle.Turtle import *

initializeTurtle(initial_speed=13)
bgcolor('black') # Základ pro vibe
colors = ['cyan', 'purple', 'white']

for i in range(120):
    color(colors[i % 3])    # Střídání 3 barev
    width(i % 10 + 1)       # Mění tloušťku čáry (Puls efekt)

    forward(i * 3)          # Stále delší čáry
    left(59)                # Nepravidelný úhel rotace
```

```python
from ColabTurtle.Turtle import *

initializeTurtle(initial_speed=13, initial_window_size=(600,500))
bgcolor('black')

colors = ['magenta', 'cyan', 'yellow', 'white']

for i in range(180):
    color(colors[i % 4])       # střídání barev
    width((i % 12) + 1)        # šířka čáry 1–12

    forward(5 + (i * 2))       # postupně delší čáry
    left(91)                   # téměř pravý úhel → krásný chaos

    # malé náhodné vlnění pohybu
    if i % 10 == 0:
        left(10)
    if i % 15 == 0:
        right(20)
```

```python
from ColabTurtle.Turtle import *
import colorsys  # pro plynulé HSV barvy

initializeTurtle(initial_speed=13, initial_window_size=(700,500))
bgcolor('black')

penup()
goto(200, 250)  # start uprostřed
pendown()

angle = 89
n = 300  # počet kroků spirály

for i in range(n):
    # plynulá změna barvy: HSV -> RGB
    hue = (i / n)  # 0–1
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)
    color(int(r * 255), int(g * 255), int(b * 255))  # nastav barvu

    width((i % 15) + 1)  # „dýchající“ tloušťka
    forward(i * 2)
    right(angle)

    # deformace spirály
    if i % 25 == 0:
        angle += 0.3
```

```python
from ColabTurtle.Turtle import *
import colorsys
import time

initializeTurtle(initial_speed=13, initial_window_size=(700,500))  # nastav rychlost v povoleném rozsahu
bgcolor('black')

penup()
goto(200, 250)
pendown()

angle = 89
n = 400  # víc kroků → delší trvání

for i in range(n):
    # plynulé barvy
    hue = (i * 2 / n) % 1
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)
    color(int(r * 255), int(g * 255), int(b * 255))

    # pulzující tloušťka
    width(1 + abs((i % 20) - 10))

    # menší krok → spirála se kreslí pomalu
    forward(1.5 * i)
    right(angle)

    # deformace spirály
    if i % 20 == 0:
        angle += 0.5
    if i % 30 == 0:
        angle -= 0.3

    time.sleep(0.01)  # krátká pauza mezi kroky → delší animace
```

```python
from ColabTurtle.Turtle import *
import colorsys
import time

initializeTurtle(initial_speed=7, initial_window_size=(700,500))
bgcolor('black')

penup()
goto(350, 250)  # start uprostřed
pendown()

n = 250          # počet kroků
angle = 144      # hvězdový úhel (5 cípy)
hue_base = 0.75  # fialová barva (HSV)

for i in range(n):
    # neon fialová (HSV) + malé pulzování
    hue = (hue_base + (i / n) * 0.05) % 1  # drobné posuny, stále fialová
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)
    color(int(r * 255), int(g * 255), int(b * 255)) # Převedení na celá čísla

    # pulzující tloušťka
    width(1 + abs((i % 15) - 7))

    forward(i * 2)  # spirálovitý růst hvězdy
    right(angle)

    # deformace pro dynamiku
    if i % 20 == 0:
        angle += 0.5
    if i % 25 == 0:
        angle -= 0.3

    time.sleep(0.01)  # zpomalit tvorbu → vidět spirálu
```
