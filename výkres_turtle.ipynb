```python
import random
from ColabTurtle.Turtle import initializeTurtle, bgcolor, color, width, penup, goto, pendown, forward, left, right, setheading
import colorsys # For generating random vibrant colors
import math # Required for spiral calculations

# Function to draw a star
def draw_star(size, x, y, star_color, star_width):
    penup()
    goto(x, y)
    pendown()
    color(star_color)
    width(star_width)
    setheading(random.randint(0, 360)) # Random orientation for the star
    for _ in range(5):
        forward(size)
        right(144) # Angle for a 5-pointed star
    penup()

# Function to draw a dot (simulated with a very short line and wide pen)
def draw_dot(x, y, dot_size, dot_color):
    penup()
    goto(x, y)
    pendown()
    color(dot_color)
    width(dot_size)
    forward(1) # Draw a very short line to simulate a dot
    penup()

# Function to draw a black hole-like spiral
def draw_black_hole_spiral(center_x, center_y, max_radius, num_steps):
    penup()
    goto(center_x, center_y)
    pendown()

    initial_angle = 0 # Starting angle
    angle_increment = 5 # How much the angle changes per step
    spiral_tightness = 0.5 # Controls how fast the spiral expands
    color_base_hue = random.random() # Random starting color for the spiral

    for i in range(num_steps):
        # Dynamic color change (circular hue shift)
        hue = (color_base_hue + (i / num_steps) * 0.5) % 1
        r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)
        color(int(r * 255), int(g * 255), int(b * 255))

        # Pulsating line width - Increased multiplier for thicker spiral
        current_width = int(1 + abs(math.sin(i * 0.1)) * 10) # Oscillates between 1 and 11
        width(current_width)

        # Calculate current radius for Archimedean spiral
        current_radius = spiral_tightness * i

        # Ensure the spiral doesn't go beyond max_radius significantly
        if current_radius > max_radius:
            current_radius = max_radius

        # Calculate new position using polar coordinates (radius, angle)
        # Angle in degrees for math.radians
        x_relative = math.cos(math.radians(initial_angle + i * angle_increment)) * current_radius
        y_relative = math.sin(math.radians(initial_angle + i * angle_increment)) * current_radius

        # Translate to screen coordinates
        screen_x = int(center_x + x_relative)
        screen_y = int(center_y + y_relative)

        goto(screen_x, screen_y)

# Initialize ColabTurtle
window_width = 800
window_height = 500
initializeTurtle(initial_speed=13, initial_window_size=(window_width, window_height))
bgcolor('black') # Set a black background

# --- Draw the black hole spiral first (as background) ---
center_x_spiral = window_width / 2
center_y_spiral = window_height / 2
max_radius_spiral = min(window_width, window_height) / 3 # About 1/3 of the smaller dimension
num_spiral_steps = 200 # Number of segments for the spiral
draw_black_hole_spiral(center_x_spiral, center_y_spiral, max_radius_spiral, num_spiral_steps)

# --- Draw random small dots ---
num_dots_to_draw = 300
min_dot_size = 1
max_dot_size = 4
for _ in range(num_dots_to_draw):
    dot_x = random.randint(0, window_width)
    dot_y = random.randint(0, window_height)
    dot_size = random.randint(min_dot_size, max_dot_size)

    # Generate random vibrant color for dots
    hue = random.random()
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)
    dot_color_rgb = (int(r * 255), int(g * 255), int(b * 255))

    draw_dot(x=dot_x, y=dot_y, dot_size=dot_size, dot_color=dot_color_rgb)

# --- Draw random smaller stars ---
num_stars_small = 20
min_star_size_small = 10
max_star_size_small = 30
for _ in range(num_stars_small):
    star_x = random.randint(20, window_width - 20)
    star_y = random.randint(20, window_height - 20)
    star_size = random.randint(min_star_size_small, max_star_size_small)

    hue = random.random()
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)
    star_color_rgb = (int(r * 255), int(g * 255), int(b * 255))

    draw_star(size=star_size, x=star_x, y=star_y, star_color=star_color_rgb, star_width=random.randint(1, 2))

# --- Draw random larger stars ---
num_stars_large = 7
min_star_size_large = 40
max_star_size_large = 80
for _ in range(num_stars_large):
    star_x = random.randint(30, window_width - 30)
    star_y = random.randint(30, window_height - 30)
    star_size = random.randint(min_star_size_large, max_star_size_large)

    hue = random.random()
    r, g, b = colorsys.hsv_to_rgb(hue, 1, 1)
    star_color_rgb = (int(r * 255), int(g * 255), int(b * 255))

    draw_star(size=star_size, x=star_x, y=star_y, star_color=star_color_rgb, star_width=random.randint(2, 4))
```
