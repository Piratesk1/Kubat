```python
class Letadlo:
    def __init__(self, cislo, typ, max_rychlost):
        self.cislo = cislo
        self.typ = typ
        self.max_rychlost = max_rychlost
        self.rychlost = 0

    def vypis_info(self):
        return f"Letadlo {self.cislo}: {self.typ}, max {self.max_rychlost} km/h, aktuální {self.rychlost} km/h"

    def zrychli(self, o_kolik):
        nova_rychlost = self.rychlost + o_kolik

        if nova_rychlost > self.max_rychlost:
            print(f"⚠️ {self.cislo}: Překročena konstrukční rychlost!")
            self.rychlost = self.max_rychlost
        else:
            self.rychlost = nova_rychlost

        return self.rychlost


l1 = Letadlo("OK-ABC", "Boeing 737", 850)
l2 = Letadlo("OK-XYZ", "Airbus A320", 830)


l1.zrychli(500)
l2.zrychli(900)

print(l1.vypis_info())
print(l2.vypis_info())
```

```python
class Letadlo:
    def __init__(self, cislo, max_rychlost):
        self.cislo = cislo
        self.max_rychlost = max_rychlost
        self.rychlost = 0

    def zrychli(self, o_kolik):
        nova_rychlost = self.rychlost + o_kolik

        if nova_rychlost > self.max_rychlost:
            print(f"⚠️ {self.cislo}: Překročena konstrukční rychlost!")
            self.rychlost = self.max_rychlost
        else:
            self.rychlost = nova_rychlost

    def vypis_info(self):
        return f"Letadlo {self.cislo}, aktuální rychlost {self.rychlost} km/h"


class VrtuloveLetadlo(Letadlo):
    def __init__(self, cislo, max_rychlost, pocet_vrtuli):
        super().__init__(cislo, max_rychlost)
        self.pocet_vrtuli = pocet_vrtuli

    def vypis_info(self):
        return f"Vrtulové letadlo {self.cislo}, vrtule: {self.pocet_vrtuli}, rychlost {self.rychlost} km/h"


class TryskoveLetadlo(Letadlo):
    def __init__(self, cislo, max_rychlost, pocet_motoru):
        super().__init__(cislo, max_rychlost)
        self.pocet_motoru = pocet_motoru

    def vypis_info(self):
        return f"Tryskové letadlo {self.cislo}, motory: {self.pocet_motoru}, rychlost {self.rychlost} km/h"


l1 = VrtuloveLetadlo("OK-123", 400, 2)
l2 = TryskoveLetadlo("OK-999", 900, 2)

l1.zrychli(300)
l2.zrychli(950)

print(l1.vypis_info())
print(l2.vypis_info())
```

```python
class Letadlo:
    def __init__(self, cislo, max_rychlost):
        self.cislo = cislo
        self.max_rychlost = max_rychlost
        self.rychlost = 0

    def zrychli(self, o_kolik):
        nova_rychlost = self.rychlost + o_kolik

        if nova_rychlost > self.max_rychlost:
            print(f"⚠️ {self.cislo}: Překročena konstrukční rychlost!")
            self.rychlost = self.max_rychlost
        else:
            self.rychlost = nova_rychlost

    def typ(self):
        return "Obecné letadlo"


# 1. úroveň dědičnosti
class TryskoveLetadlo(Letadlo):
    def __init__(self, cislo, max_rychlost, pocet_motoru):
        super().__init__(cislo, max_rychlost)
        self.pocet_motoru = pocet_motoru

    def typ(self):
        return f"Tryskové ({self.pocet_motoru} motory)"


# 2. úroveň dědičnosti (podtřída podtřídy!)
class BojoveLetadlo(TryskoveLetadlo):
    def __init__(self, cislo, max_rychlost, pocet_motoru, zbrane):
        super().__init__(cislo, max_rychlost, pocet_motoru)
        self.zbrane = zbrane

    def typ(self):
        return f"Bojové letadlo, zbraně: {self.zbrane}"


# další větev
class VrtuloveLetadlo(Letadlo):
    def __init__(self, cislo, max_rychlost, pocet_vrtuli):
        super().__init__(cislo, max_rychlost)
        self.pocet_vrtuli = pocet_vrtuli

    def typ(self):
        return f"Vrtulové ({self.pocet_vrtuli} vrtule)"


# objekty různých úrovní
l1 = Letadlo("OK-000", 300)
l2 = TryskoveLetadlo("OK-111", 900, 2)
l3 = BojoveLetadlo("OK-222", 1200, 2, "rakety")
l4 = VrtuloveLetadlo("OK-333", 400, 2)

# polymorfismus + dědičnost
letadla = [l1, l2, l3, l4]

for l in letadla:
    l.zrychli(500)
    print(f"{l.cislo}: {l.typ()}, rychlost {l.rychlost} km/h")
```
