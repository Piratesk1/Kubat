```python
import statistics as stats
import matplotlib.pyplot as plt

# ==========================================
# DATOVÁ SADA LETADEL
# (název, rychlost km/h, dolet km, rok)
# ==========================================

letadla = [
    ("Boeing 737", 876, 6570, 1967),
    ("Boeing 747", 933, 14100, 1969),
    ("Airbus A320", 871, 6300, 1987),
    ("Airbus A380", 1020, 15400, 2005),
    ("Concorde", 2170, 7250, 1969),
    ("F-16 Falcon", 2400, 4220, 1974),
    ("Cessna 172", 226, 1289, 1955),
    ("MiG-21", 2230, 1210, 1955),
]

# ==========================================
# FUNKCE
# ==========================================

def vypis(data, nadpis="Letadla"):
    print(f"\n--- {nadpis} ---")
    for l in data:
        print(f"{l[0]} | {l[1]} km/h | {l[2]} km | {l[3]}")
    print()

def statistiky(data):
    rychlosti = [l[1] for l in data]
    dolet = [l[2] for l in data]

    print("\n--- STATISTIKY ---")
    print(f"Průměrná rychlost: {stats.mean(rychlosti):.1f} km/h")
    print(f"Nejvyšší rychlost: {max(rychlosti)} km/h")
    print(f"Nejnižší rychlost: {min(rychlosti)} km/h")

    print(f"\nPrůměrný dolet: {stats.mean(dolet):.1f} km")
    print(f"Největší dolet: {max(dolet)} km")
    print(f"Nejmenší dolet: {min(dolet)} km\n")

def histogram_rychlosti(data):
    rychlosti = [l[1] for l in data]

    plt.hist(rychlosti, bins=8, edgecolor="black")
    plt.title("Histogram rychlostí letadel")
    plt.xlabel("Rychlost (km/h)")
    plt.ylabel("Počet letadel")
    plt.show()

def graf_doletu(data):
    nazvy = [l[0] for l in data]
    dolet = [l[2] for l in data]

    plt.figure(figsize=(10, 6))
    plt.barh(nazvy, dolet, color="skyblue")
    plt.title("Dolet letadel")
    plt.xlabel("Dolet (km)")
    plt.ylabel("Letadlo")
    plt.tight_layout()
    plt.show()

# ==========================================
# START – AUTOMATICKÝ VÝSTUP
# ==========================================

# 1) Výpis všech letadel
vypis(letadla, "Všechny letadla")

# 2) Seřazeno podle rychlosti
vypis(sorted(letadla, key=lambda x: x[1], reverse=True), "Seřazeno podle rychlosti")

# 3) Seřazeno podle doletu
vypis(sorted(letadla, key=lambda x: x[2], reverse=True), "Seřazeno podle doletu")

# 4) Seřazeno podle roku prvního letu
vypis(sorted(letadla, key=lambda x: x[3]), "Seřazeno podle roku prvního letu")

# 5) Statistické údaje
statistiky(letadla)

# 6) Histogram rychlostí
histogram_rychlosti(letadla)

# 7) Sloupcový graf doletu
graf_doletu(letadla)
```
