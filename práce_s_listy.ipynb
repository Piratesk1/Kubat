```python
import json
import statistics as stats
import matplotlib.pyplot as plt


# ============================================================
#                DATASET LETECKÝCH STROJŮ
#  (název, maximální rychlost km/h, dolet km, rok prvního letu)
# ============================================================

letadla = [
    ("Boeing 737", 876, 6570, 1967),
    ("Boeing 747", 933, 14100, 1969),
    ("Boeing 777", 950, 15600, 1994),
    ("Boeing 787", 913, 13620, 2009),
    ("Airbus A320", 871, 6300, 1987),
    ("Airbus A330", 913, 13450, 1992),
    ("Airbus A350", 945, 15000, 2013),
    ("Airbus A380", 1020, 15400, 2005),
    ("Concorde", 2170, 7250, 1969),
    ("F-16 Falcon", 2400, 4220, 1974),
    ("F-35 Lightning II", 1930, 2220, 2006),
    ("MiG-29", 2450, 1430, 1977),
    ("MiG-21", 2230, 1210, 1955),
    ("Cessna 172", 226, 1289, 1955),
    ("Bombardier Q400", 667, 2500, 1998),
    ("Antonov An-225", 850, 15400, 1988),
]


# ============================================================
#                        FUNKCE
# ============================================================

def vypis_letadla(data):
    print("\n--- Přehled letadel ---")
    for l in data:
        print(f"{l[0]} | Rychlost: {l[1]} km/h | Dolet: {l[2]} km | Rok: {l[3]}")
    print()


def filtr_podle_rychlosti():
    min_rychlost = int(input("Zadej minimální rychlost (km/h): "))
    vysledek = [l for l in letadla if l[1] >= min_rychlost]
    vypis_letadla(vysledek)


def filtr_podle_doletu():
    min_dolet = int(input("Zadej minimální dolet (km): "))
    vysledek = [l for l in letadla if l[2] >= min_dolet]
    vypis_letadla(vysledek)


def filtr_podle_roku():
    rok = int(input("Zadej minimální rok prvního letu: "))
    vysledek = [l for l in letadla if l[3] >= rok]
    vypis_letadla(vysledek)


def vyhledat():
    text = input("Zadej část názvu letadla: ").lower()
    vysledek = [l for l in letadla if text in l[0].lower()]
    vypis_letadla(vysledek)


def serad_podle_nazvu():
    vypis_letadla(sorted(letadla, key=lambda x: x[0]))


def serad_podle_rychlosti():
    vypis_letadla(sorted(letadla, key=lambda x: x[1], reverse=True))


def serad_podle_doletu():
    vypis_letadla(sorted(letadla, key=lambda x: x[2], reverse=True))


def serad_podle_roku():
    vypis_letadla(sorted(letadla, key=lambda x: x[3]))


def statistiky():
    rychlosti = [l[1] for l in letadla]
    dolet = [l[2] for l in letadla]

    print("\n--- STATISTIKY ---")
    print(f"Průměrná rychlost: {stats.mean(rychlosti):.1f} km/h")
    print(f"Nejrychlejší: {max(rychlosti)} km/h")
    print(f"Nejpomalejší: {min(rychlosti)} km/h")
    print()
    print(f"Průměrný dolet: {stats.mean(dolet):.1f} km")
    print(f"Největší dolet: {max(dolet)} km")
    print(f"Nejmenší dolet: {min(dolet)} km\n")


def histogram_rychlosti():
    rychlosti = [l[1] for l in letadla]
    plt.hist(rychlosti, bins=10, edgecolor="black")
    plt.title("Histogram maximálních rychlostí letadel")
    plt.xlabel("Rychlost (km/h)")
    plt.ylabel("Počet letadel")
    plt.show()


def bar_dolet():
    nazvy = [l[0] for l in letadla]
    dolet = [l[2] for l in letadla]

    plt.figure(figsize=(12, 6))
    plt.barh(nazvy, dolet)
    plt.title("Dolet letadel")
    plt.xlabel("Dolet (km)")
    plt.ylabel("Letadlo")
    plt.tight_layout()
    plt.show()


def porovnat():
    print("Zadej názvy dvou letadel k porovnání.")
    l1 = input("Letadlo 1: ")
    l2 = input("Letadlo 2: ")

    a = next((x for x in letadla if x[0].lower() == l1.lower()), None)
    b = next((x for x in letadla if x[0].lower() == l2.lower()), None)

    if not a or not b:
        print("Jedno nebo obě letadla nebyla nalezena.")
        return

    print("\n--- POROVNÁNÍ ---")
    print(f"{a[0]} vs {b[0]}")
    print(f"Rychlost: {a[1]} vs {b[1]}")
    print(f"Dolet: {a[2]} vs {b[2]}")
    print(f"Rok: {a[3]} vs {b[3]}\n")


def pridat_letadlo():
    název = input("Název letadla: ")
    rychlost = int(input("Maximální rychlost km/h: "))
    dolet_ = int(input("Dolet km: "))
    rok = int(input("Rok prvního letu: "))

    letadla.append((název, rychlost, dolet_, rok))
    print("Letadlo úspěšně přidáno!")


def ulozit_json():
    with open("letadla.json", "w", encoding="utf-8") as f:
        json.dump(letadla, f, ensure_ascii=False, indent=4)
    print("Uloženo do letadla.json")


def nacist_json():
    global letadla
    try:
        with open("letadla.json", "r", encoding="utf-8") as f:
            letadla = [tuple(x) for x in json.load(f)]
        print("Načteno!")
    except:
        print("Nepodařilo se načíst soubor.")


# ============================================================
#                         MENU
# ============================================================

def menu():
    while True:
        print("\n===============================")
        print("     APLIKACE – LETADLA ✈️")
        print("===============================")
        print("1) Výpis letadel")
        print("2) Vyhledávání podle názvu")
        print("3) Filtr podle rychlosti")
        print("4) Filtr podle doletu")
        print("5) Filtr podle roku")
        print("6) Řazení podle názvu")
        print("7) Řazení podle rychlosti")
        print("8) Řazení podle doletu")
        print("9) Řazení podle roku")
        print("10) Statistické údaje")
        print("11) Histogram rychlostí")
        print("12) Sloupcový graf doletu")
        print("13) Porovnání dvou letadel")
        print("14) Přidat letadlo")
        print("15) Uložit do JSON")
        print("16) Načíst z JSON")
        print("0) Konec")
        print("===============================")

        volba = input("Vyber možnost: ")

        akce = {
            "1": lambda: vypis_letadla(letadla),
            "2": vyhledat,
            "3": filtr_podle_rychlosti,
            "4": filtr_podle_doletu,
            "5": filtr_podle_roku,
            "6": serad_podle_nazvu,
            "7": serad_podle_rychlosti,
            "8": serad_podle_doletu,
            "9": serad_podle_roku,
            "10": statistiky,
            "11": histogram_rychlosti,
            "12": bar_dolet,
            "13": porovnat,
            "14": pridat_letadlo,
            "15": ulozit_json,
            "16": nacist_json,
        }

        if volba == "0":
            print("Ukončuji aplikaci…")
            break

        if volba in akce:
            akce[volba]()
        else:
            print("Neplatná volba.")


# ============================================================
#                       START PROGRAMU
# ============================================================

if __name__ == "__main__":
    menu()
```
